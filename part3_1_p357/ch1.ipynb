# 가설 검정

## 왜 가설검정이 필요할까??
- 모집단 전체를 전수조사할 수 있다면 가설검정은 필요가 없다.
- 비용, 시간 등의 문제로 모집단의 일부인 "표번(sample)"만 관찰할 수 있을 때, 그 표본을 근거로 모집단에 대한 결론을 통계적으로 추론하기 위해 가설검정을 사용한다.
- 작업형 3번 시험에서 주어지는 데이터는 항상 모집단에서 뽑은 표본 데이터라고 가정하고 문제를 푼다.

## 귀무가설 vs 대립가설

### 귀무가설
- 기존에 알려진 사실, 변화, 차이가 "없다"라는 주장

### 대립가설
- 새로 입증하고자 하는 주장, 변화, 차이가 "있다"는 주장
- 예) 아메리카노 한 잔의 원두는 20g이 아니다.

- 검정 결과를 해석할 때 통계학에서는 귀무가설이 옳다고 발쳐져도 "채택한다"보다는 "기각하지 못한다"라는 표현을 정확한 용어로 사용한다.
- 시험 답안 작성 시에서는 "귀무가설 채택" 또는 "귀무가설 기각" 처럼 명확하게 답한다.

- p-value < 유의수준 : 귀무가설 기각, 대립가설 채택
- p_value >= 유의수준 : 귀무가설 채택, 대립가설 기각

### 가설검정 프로세스 4단계 (교재 pdf에 있다!)
1. 통계적 가설 설정 : 귀무가설과 대립가설을 정의
2. 유의수준 결정 : 보통 0.05 또는 0.01(문제에서 특별히 안내하지 않으면 0.05가 기본)
3. 검정통계량 및 유의확률(p-value) 계산 : scipy.stats 각 검정 함수 사용
4. 결과 도출 : p-value와 유의수준을 비교해 기각/채택 결론 도출


## 1. 단일 표본검정 (One-sample t-test)
- 이 표본의 평균이 특정 기준값과 통계적으로 다른가?? 를 검정
- 예) 영화관 팝콘 라지 사이즈의 평균 무게가 알려진 120g과 같은지 확인

```python
import pandas as pd

# 30명의 몸무게 데이터 (weights 컬럼 하나만 있는 데이터프레임)
df = pd.DataFrame({
    'weights':[122, 121, 120, 119, 125, 115, 121, 118, 117, 127,
           123, 129, 119, 124, 114, 126, 122, 124, 121, 116,
           120, 123, 127, 118, 122, 117, 124, 125, 123, 121],
})
```

```python
df.head()
```

### ttest_lsamp(a, popmean, alternative)
- a: 모집단에서 뽑은 표본 데이터
- popmean: 비교하려는 모집단의 평균(기대값). 그리스문자 뮤 제로로 표기한다.
    - 뮤 : 표본이 속한 모집단의 실제 평균
    - 뮤 제로 : 귀무가설에서 주장하는 (기존에 알려진) 평균값

- alternative : 대립가설의 방향을 정의
    - 'greater'p > 뮤 제로 : 표본 평균이 기준값보다 크다.
    - 'less'p < 뮤 제로 : 표본 평균이 기준값보다 작다.
    - 'two-sided' 기본값, 표본 평균이 기준값과 다르다. 

```python
from scipy import stats

t_statistic, p_value = stats.ttest_1samp(df['weights'], 120)
print(f't_statistic: {t_statistic}')
print(f'p_value: {p_value}')
```

```python
# 양측 검정 - 기본값
print(stats.ttest_1samp(df['weights'], 120, alternative='two-sided'))
```

```python
# 단특 검정 - 오른쪽 --> 평균이 120보다 크다를 대립가설로 검정
print(stats.ttest_1samp(df['weights'], 120, alternative='greater'))
```

```python
# # 단특 검정 - 왼쪽 --> 평균이 120보다 크다를 대립가설로 검정
print(stats.ttest_1samp(df['weights'], 120, alternative='less'))
```

#### 주의할점
- 같은 데이터라도 어떤 alternative를 사용하느냐에 따라 p-value가 달라지고, 결론(기각/채택)이 뒤바뀔 수 있다.
- 문제 지문의 방향성을 정확히 체크해야 한다.

### 비모수 검정 - 정규성이 꺠졌을 때
- 정규성 검정에는 "귀무가설 채택" 기대한다.
- 일반적인 가설검정(t-검정 등)에서는 연구자가 대립가설의 채택(=유의미한 차이 발견)을 기대하는 경우가 많다.
- 데이터가 정규분포를 "따른다"는 귀무가설이 채택되기를 기대하는 것이 일반적이다.
- 그래야 모수적 방법인 t-검정을 그대로 사용할 수 있기 때문이다.

- 귀무가설(뮤 제로) : 데이터가 정규분포를 따른다
- 대립가설(뮤 원) : 데이터가 정규분포를 따르지 않는다.

```python
import pandas as pd 
df = pd.DataFrame({ 
    'weights' :[1125, 126, 118, 124, 117, 127, 123, 122, 119, 142]
})
```

```python
# 정규성이 깨졌으므로 t-검정 대신 Wilcoxon 부호순위 검정(비모수) 사용
# 각 값에서 기준값 120을 뼨 '차이값'을 검정 대상으로 사용
stats.wilcoxon(df['weights'] - 120, alternative='less')
```

- 단일표본 비모수 검정에는 데이터 - 기준값으로 차이를 만둔 뒤 wilcoxon()에 넣는 패턴을 기억해두자!
- wilcoxon 검정은 평균이 아닌 중앙값에 대한 가설을 검증한다

## 대응 표본 검정 (Paired t-test)

### 대응 표본 검정
- 모집단 1개 (같은 집단을 두 번 측정)
- 같은 그룹의 두 시점(조건) 비교
- 예) 체중 감량 프로그램 (전/후), 교육 프로그램(전/후)

```python
import pandas as pd
# 같은 사람들의 '전(before)'과 '후(after)' 측정값 → 서로 짝지어진(paired) 데이터
df = pd.DataFrame({
    'before':[85, 90, 92, 88, 86, 89, 83, 87],
    'after':[85.5,89.9,92.6,89.5,85.8,88.8,84.6,87.8]
})

df.head()
```

```python
from scipy import stats

# alternative='less'
# before - after의 평균이 0보다 작다 --> 검정
#   before < after
stats.ttest_rel(df['before'], df['after'], alternative='less')

# alternative='greater'
# before - after의 평균이 0보다 크다 --> 검정
#   before < after
stats.ttest_rel(df['after'], df['before'], alternative='greater')
```

- 대응 표본 검정 문제는 "효과가 있는지"를 묻는 문제가 많다.
- 양측 검정보다 단측 검정을 사용하는 빈도가 더 높다.
    - 사전보다 사후가 개선되었는지를 확인하는 방향성있는 질문이 대부분이기 때문.

### 비모수 검정 - 정규성이 깨졌을 때

```python
import pandas as pd
df = pd.DataFrame({
    'before':[85, 90, 92, 88, 86, 89, 83, 87],
    'after':[86, 92, 94, 89, 84, 90, 84, 88]
})
```

```python
from scipy import stats

# 두 그룹의 차이를 새 컬럼으로 만든다 -> 그 차이값이 정규분포를 따르는지 검정
df['diff'] = df['after'] - df['before']
stats.shapiro(df['diff'])
```

- t-검정 할 때는 순서가 결과 해석에 영향을 준다.
- shapiro()에 넣는 데이터 순서는 diff를 구했기 때문에 결과 해석에 영향을 주지 않는다.

```python
stats.wilcoxon(df['after'], df['before'], alternative='greater')
```

```python
stats.wilcoxon(df['diff'], alternative='greater')   # 위와 같다.
```

### 독립 표본 검정
- 두 그룹(표본)간의 평균이 서로 다름을 판단하는 통계 방법
- A그룹과 B그룹을 나누고, 각 그룹을 대상으로 적절한 통계쩍 검정을 통해 두 그룹 사이에 통계적으로 유의미한 차이가 있는지 없는지 결론을 내리는 것 
- "통계적으로 유의미하다"라는 표현을 사용하는 이유 : 우현히 발생할 수 있는 차이와 실제로 의미 있는 차이를 구분하기 위해서이다.

### ttest_ind(첫 번째 모집단에서 뽑은 표본 데이터, 두 번째 모집단에서 뽑은 표본 데이터, 대립가설 정의, 두 모집단의 분산), 두 모집단의 분산-같다(기본값))

- 양측 검정
    - 귀무가설 : 반별 수학 평균 점수는 같다.
    - 대립가설 : 반별 수학 평균 점수는 다르다.


```python
import pandas as pd
# 서로 다른 두 반의 점수 (표본 크기가 달라도 상관없음: 8명 vs 5명)
class1 = [85, 90, 92, 88, 86, 89, 83, 87]
class2 = [80, 82, 88, 85, 84]
```

```python
from scipy.stats import ttest_ind

ttest_ind(class1, class2)
```

```python
# 모분산이 다르다면 equal_val = False
ttest_ind(class1, class2, equal_var=False)
```

- 단축 검정 (전제조건. 모분산은 동일히다)
    - 귀무가설 : 반별 수학 평균 점수는 같다.
    - 대립가설 : 2반 수학 평균 점수가 더 높다. (1번 수학 평균 점수가 더 낮다.)

```python
ttest_ind(class1, class2, alternative='less', equal_var = True)
```

- 단축 검정 (전제조건. 모분산은 동일히다)
    - 귀무가설 : 반별 수학 평균 점수는 같다.
    - 대립가설 : 1반 수학 평균 점수가 더 높다. (2번 수학 평균 점수가 더 낮다.)

```python
ttest_ind(class1, class2, alternative='greater', equal_var = True)
```

### 비모수 검정
- 정규성에 위배
- 샤피로-윌크 검정을 먼저 수행 -> 정규성 가정이 위배되면 -> 맨-휘트니U 검정을 수행

```python
import pandas as pd
class1 = [85, 90, 92, 88, 86, 89, 83, 87]
class2 = [80, 82, 88, 85, 130]
```

```python
from scipy import stats

stats.shapiro(class1)
```

```python
stats.shapiro(class2)
```

```python
# class2가 정규성을 만족하지 않았다 --> p < 0.05 --> t- 검정 대신 Mann-Whitney U 검정(비모수, 독립표본)
# "class1이 class2보다 작다"를 검정
stats.mannwhitneyu(class1, class2, alternative='less')
```

- 독립표본에서 정규성이 깨지면 --> Mann_Whitney U 검정
- 대응표본에서 정규성이 깨지면 --> Wilcoxon 검정

```python
result = stats.mannwhitneyu(class1, class2, alternative='less').pvalue

# format(값, '.소수자리수f')
format(result, '.2f')   # 소수 둘째자리까지 반올림
```
