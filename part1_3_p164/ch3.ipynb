### SECTION 03 - 파생변수, 정렬, 인덱싱 (p.168)

```python
import pandas as pd
df3 = pd.read_csv('type1_data1.csv')
```

```python
df3 = df3.dropna()
df3.shape
```

```python
df3['new'] = df3['views']/df3['f1']
df3.head()
```

```python
# df3['new'].sort_Values(ascending=False) -> new 컬럼 값만 담긴 Series만 남아 age이 결과에 없어짐
df3 = df3.sort_values('new', ascending=False)
df3.head()
```

```python
int(df3.iloc[0,1])
```

### SECTION 04 - 값 변경, 정렬, 합계(p.169)

```python
df4 = pd.read_csv('type1_data1.csv')
```

```python
df4['views'] = df4['views'].fillna(0)
```

```python
value = df4['views'].nlargest(10).iloc[-1]
value
```

```python
df4.loc[df4['views']>value,'views'] = value
print(int(df4['views'].sum()))
```
