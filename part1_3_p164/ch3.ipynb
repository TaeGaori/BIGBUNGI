### SECTION 16 - 이상치, IQR(p.185)

```python
df16 = pd.read_csv('type1_data1.csv')
```

```python
q3 = df16['views'].quantile(.75)
q1 = df16['views'].quantile(.25)

IQR = q3 - q1
IQR
```

```python
line1 = q1 - IQR * 1.5
line2 = q3 + IQR * 1.5
```

```python
cond16_1 = df16['views'] < line1
cond16_2 = df16['views'] > line2

df16 = df16[cond16_1 | cond16_2]
```

```python
int(df16['views'].sum())
```
